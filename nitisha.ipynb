```python
class Smartphone:
    def __init__(self, brand, model):
        self.brand = brand           
        self.model = model           
        self.__battery = 100         

    
    def get_battery(self):
        return f"{self.__battery}%"
def use_phone(self, minutes):
        drain = minutes * 1  
        if self.__battery - drain >= 0:
            self.__battery -= drain
        else:
            self.__battery = 0
        print(f"Used phone for {minutes} mins. Battery is now {self.__battery}%.")
def charge(self, amount):
        if amount > 0:
            self.__battery = min(100, self.__battery + amount)
            print(f"Charged! Battery is now {self.__battery}%.")

# --- Example Usage ---
my_phone = Smartphone("Samsung", "Galaxy S24")

print(my_phone.brand)          
print(my_phone.model)          
print(my_phone.get_battery())  





    
     
```

```python
# Initial list of systolic blood pressure readings (in mmHg)
bp_readings = [120, 130, 140, 120, 150, 135, 145, 125, 138, 128]

# Remove a duplicate reading of 120 mmHg
bp_readings.remove(120)
print("After removing duplicate (120 mmHg):", bp_readings)

# Remove the first reading (e.g., faulty entry reading)
bp_readings.pop(0)
print("After removing first reading:", bp_readings)

# Delete the first reading (another method, e.g., repeat entry correction)
del bp_readings[0]
print("After deleting first reading:", bp_readings)

# Clear all readings for a new patient or new day
bp_readings.clear()
print("After clearing all readings:", bp_readings)
```

```python
# Create a set with duplicate values
numbers = {10, 20, 30, 20, 40, 10}
# Duplicates are automatically removed
print("Set:", numbers)

# Add one element
numbers.add(50)
print("After add:", numbers)

# Add multiple elements
numbers.update({60, 70, 80})
print("After update:", numbers)

# Remove an element if it exists
numbers.discard(100)  # No error if 100 is not present
numbers.discard(20)
print("After discard:", numbers)
```
