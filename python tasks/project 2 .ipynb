```python
# 1 Load and inspect the dataset
import pandas as pd

# Load the CSV (keep it in the same folder as your notebook)
df = pd.read_csv("02_student_performance (4).csv")

# Inspect the dataset
display(df.head())        # First 5 rows
print("Shape:", df.shape) # Rows and columns
print("Columns:", df.columns.tolist())
df.info()                 # Column types and non-null counts
display(df.describe())    # Summary statistics for numeric columns




```

```python
# 2 Report missing values and duplicate Student_ID values
# Missing values by column
print("Missing values per column:")
print(df.isna().sum())

# Repeated Student_ID values
duplicate_ids = df.loc[
    df["Student_ID"].duplicated(keep=False), "Student_ID"
].value_counts()

print("\nDuplicate Student_ID values and their counts:")
if duplicate_ids.empty:
    print("No duplicate Student_ID values found.")
else:
    print(duplicate_ids)
```

```python

print(df.columns.tolist())
```

```python
# 3 Handle missing attendance values and explain the chosen method
# Median keeps the fill value balanced if there are unusually low or high attendance values
att_col = "Attendance_Percent"

missing_before = df[att_col].isna().sum()
attendance_median = df[att_col].median()

df[att_col] = df[att_col].fillna(attendance_median)

print("Missing attendance values before:", missing_before)
print("Median used:", attendance_median)
print("Missing attendance values after:", df[att_col].isna().sum())
```

```python
# 4 Check score columns for values outside 0–100.
score_columns = [
    col for col in df.columns
    if "mark" in col.lower() or "score" in col.lower()
]

if not score_columns:
    print("No score columns found. Check the column names with df.columns.tolist().")
else:
    for col in score_columns:
        scores = pd.to_numeric(df[col], errors="coerce")
        outside_range = (scores < 0) | (scores > 100)

        print(f"\n{col}: {outside_range.sum()} value(s) outside 0–100")
        if outside_range.any():
            display(df.loc[outside_range, [col]])
```

```python

# 5 Create a Grade column using clearly defined score ranges.
# Pick subject mark columns (excluding total/average columns so they don't skew the mean)
score_columns = [
    col for col in df.columns
    if ("mark" in col.lower() or "score" in col.lower())
    and "average" not in col.lower()
    and "total" not in col.lower()
]

if not score_columns:
    print("No subject score columns found. Check with df.columns.tolist().")
else:
    scores = df[score_columns].apply(pd.to_numeric, errors="coerce")
    df["Average_Marks"] = scores.mean(axis=1)

    df["Grade"] = pd.cut(
        df["Average_Marks"],
        bins=[-float("inf"), 60, 70, 80, 90, float("inf")],
        labels=["F", "D", "C", "B", "A"],
        right=False,
    )

    print("Score columns used:", score_columns)
    print(df["Grade"].value_counts().sort_index())
    display(df[["Average_Marks", "Grade"]].head())
```

```python
# 6 Calculate department-wise average marks.
department_avg = (
    df.groupby("Department")["Average_Marks"]
      .mean()
      .reset_index()
      .sort_values("Average_Marks", ascending=False)
)

display(department_avg)
```

```python
# 7 Find the top 10 students by average marks.
top_10_students = df.nlargest(10, "Average_Marks")
display(top_10_students)
```

```python
# 8 Identify students below 40% average marks.
below_40 = df[df["Average_Marks"] < 40]

print("Number of students below 40%:", len(below_40))
display(below_40)
```

```python
# 9 Compare average marks for different attendance bands.
df["Attendance_Band"] = pd.cut(
    df["Attendance_Percent"],
    bins=[-float("inf"), 60, 75, 90, float("inf")],
    labels=["Below 60%", "60-74%", "75-89%", "90% or more"],
    right=False,
)

attendance_comparison = (
    df.groupby("Attendance_Band", observed=True)
    .agg(
        Students=("Average_Marks", "count"),
        Average_Marks=("Average_Marks", "mean"),
    )
    .reset_index()
)

display(attendance_comparison)
```

```python
# 10 Analyze the relationship between study hours and average marks.
import matplotlib.pyplot as plt

study_col = "Study_Hours_Per_Day"

study_data = df[[study_col, "Average_Marks"]].apply(
    pd.to_numeric, errors="coerce"
).dropna()

correlation = study_data[study_col].corr(study_data["Average_Marks"])
print(f"Correlation between study hours and average marks: {correlation:.2f}")

plt.scatter(study_data[study_col], study_data["Average_Marks"])
plt.xlabel("Study Hours Per Day")
plt.ylabel("Average Marks")
plt.title("Study Hours vs Average Marks")
plt.show()
```

```python
# 11 Compare results for students with and without internet access.
internet_comparison = (
    df.groupby("Internet_Access")["Average_Marks"]
      .agg(["count", "mean", "median"])
      .rename(columns={
          "count": "Student_Count",
          "mean": "Average_Marks",
          "median": "Median_Marks"
      })
)

display(internet_comparison)
```

```python
# 12 Calculate median, standard deviation and quartiles of Python_Marks using NumPy.


import numpy as np

marks = pd.to_numeric(df["Python_Marks"], errors="coerce").dropna().to_numpy()

print("Median:", np.median(marks))
print("Standard deviation:", np.std(marks))  # Population standard deviation

q1, q2, q3 = np.percentile(marks, [25, 50, 75])
print("Q1 (25th percentile):", q1)
print("Q2 (50th percentile):", q2)
print("Q3 (75th percentile):", q3)
```

```python
# 13 Identify potential score outliers using IQR.
score_columns = [
    col for col in df.columns
    if ("mark" in col.lower() or "score" in col.lower())
    and "average" not in col.lower()
    and "total" not in col.lower()
]

if not score_columns:
    print("No score columns found. Check the names with df.columns.tolist().")
else:
    for col in score_columns:
        scores = pd.to_numeric(df[col], errors="coerce")
        q1 = scores.quantile(0.25)
        q3 = scores.quantile(0.75)
        iqr = q3 - q1
        lower_bound = q1 - 1.5 * iqr
        upper_bound = q3 + 1.5 * iqr

        outliers = scores.notna() & (
            (scores < lower_bound) | (scores > upper_bound)
        )

        print(f"\n{col}: {outliers.sum()} potential outlier(s)")
```

```python
# 15 calculate total and average marks for each student.
score_columns = [
    col for col in df.columns
    if ("mark" in col.lower() or "score" in col.lower())
    and "average" not in col.lower()
    and "total" not in col.lower()
]

scores = df[score_columns].apply(pd.to_numeric, errors="coerce")
df["Total_Marks"] = scores.sum(axis=1, min_count=1)
df["Average_Marks"] = scores.mean(axis=1)

display(df[["Student_ID", "Total_Marks", "Average_Marks"]])
```
