```python
import math

# 1. Compute subtotal, 18% GST, and final amount
unit_price = 450.0
quantity = 5
subtotal = unit_price * quantity
gst = subtotal * 0.18
final_amount = subtotal + gst

print(f"Subtotal: {subtotal}")
print(f"GST (18%): {gst}")
print(f"Final Amount: {final_amount}")
print("-" * 30)

# 2. Use // and % to express 100 notebooks as full cartons of 12 plus loose units
total_notebooks = 100
carton_size = 12

full_cartons = total_notebooks // carton_size
loose_units = total_notebooks % carton_size

print(f"Total Notebooks: {total_notebooks}")
print(f"Full Cartons (of 12): {full_cartons}")
print(f"Loose Units: {loose_units}")
print("-" * 30)

# 3. Show that 0.1 + 0.2 == 0.3 is False and fix it using round() or math.isclose()
print(f"0.1 + 0.2 == 0.3 is: {0.1 + 0.2 == 0.3}")  # False due to floating-point precision

# Fixing using round()
fixed_round = round(0.1 + 0.2, 2) == round(0.3, 2)
print(f"Fixed with round(): {fixed_round}")

# Fixing using math.isclose()
fixed_isclose = math.isclose(0.1 + 0.2, 0.3)
print(f"Fixed with math.isclose(): {fixed_isclose}")
```

```python


print("TASK 3: SLAB DISCOUNT")

GST_RATE = 0.18

orders = [
    {"order_id": 2026_10067,  "product": "AIRFRYER",        "quantity": 6,  "unit_price": 12500.65, "is_cancelled": False},
    {"order_id": 2026_10058,  "product": "ELECTRIC KETTLE", "quantity": 4,  "unit_price": 3400.78,  "is_cancelled": True},
    {"order_id": 2026_10176,  "product": "MICROWAVE OVEN",  "quantity": 16, "unit_price": 8950.50,  "is_cancelled": False},
    {"order_id": 2026_10890,  "product": "REFRIGERATOR",    "quantity": 9,  "unit_price": 27478.87, "is_cancelled": False},
    {"order_id": 2026_100098, "product": "SMART TV",        "quantity": 7,  "unit_price": 68679.0,  "is_cancelled": False},
    {"order_id": 2026_10059,  "product": "LAPTOP",          "quantity": 21, "unit_price": 56569.45, "is_cancelled": False},
    
]

for order in orders:
    order_id = order["order_id"]
    product = order["product"]
    quantity = order["quantity"]
    unit_price = order["unit_price"]

    # Valid only when BOTH conditions hold (and).
    # Same rule flipped with 'or': invalid if quantity <= 0 or unit_price <= 0
    is_valid = quantity > 0 and unit_price > 0
    if not is_valid:
        print(f"{order_id} | {product:<15} | INVALID order (quantity and price must be > 0)")
        continue

    if order["is_cancelled"]:
        print(f"{order_id} | {product:<15} | CANCELLED - not billed")
        continue

    subtotal = quantity * unit_price

    # Slab discount using if / elif / else
    if subtotal >= 10000:
        discount_rate = 0.10
    elif subtotal >= 5000:
        discount_rate = 0.05
    else:
        discount_rate = 0.0

    discount = subtotal * discount_rate
    taxable = subtotal - discount
    gst = taxable * GST_RATE
    final_amount = taxable + gst

    print(f"{order_id} | {product:<15} | Subtotal: {subtotal:>12,.2f} | "
          f"Discount ({discount_rate:.0%}): {discount:>10,.2f} | "
          f"GST: {gst:>10,.2f} | Final: {final_amount:>12,.2f}")

print("TASK 4: BILL LOOP")

GST_RATE = 0.18

# (order_id, product, quantity, unit_price, is_cancelled)
orders = [
    (2026_10067,  "hair dryer"  ,      6,  12500.65, False),
    (2026_10058,  "charging cabel", 4,  3400.78,  True),
    (2026_10176,  "earphones",  16, 500.50,  False),
    (2026_10890,  "REFRIGERATOR",    9,  27478.87, False),
    (2026_100098, "SMART watch",        7,  6869.0,  False),
    (2026_10058,  "LAPTOP",          21, 56569.45, False),
]

grand_total = 0.0
billed = 0
skipped = 0

print(f"{'#':<3}{'Order ID':<12}{'Product':<17}{'Subtotal':>14}{'Discount':>13}{'GST':>13}{'Final':>15}")
print("-" * 87)

for number, (order_id, product, quantity, unit_price, is_cancelled) in enumerate(orders, start=1):
    # Skip invalid or cancelled orders
    if quantity <= 0 or unit_price <= 0 or is_cancelled:
        print(f"{number:<3}{order_id:<12}{product:<17}-- skipped (invalid or cancelled) --")
        skipped += 1
        continue

    subtotal = quantity * unit_price

    if subtotal >= 10000:
        discount = subtotal * 0.10
    elif subtotal >= 5000:
        discount = subtotal * 0.05
    else:
        discount = 0.0

    gst = (subtotal - discount) * GST_RATE
    final_amount = subtotal - discount + gst

    print(f"{number:<3}{order_id:<12}{product:<17}{subtotal:>14,.2f}{discount:>13,.2f}{gst:>13,.2f}{final_amount:>15,.2f}")

    grand_total += final_amount
    billed += 1

print("-" * 87)
print(f"Orders billed: {billed} | Skipped: {skipped}")
print(f"GRAND TOTAL: {grand_total:,.2f}")
```

```python
print("TASK 5:INPUT VALIDATION ")

MAX_ATTEMPTS = 3

# New order for a product from the RetailPulse data
order_id = 2026_10100
product = "AIRFRYER"
unit_price = 12500.65

print(f"New order {order_id}: {product} @ {unit_price:,.2f}")

attempts = 0
quantity = None

while attempts < MAX_ATTEMPTS:
    attempts += 1
    raw = input(f"Enter quantity (attempt {attempts}/{MAX_ATTEMPTS}): ").strip()

    # isdecimal() rejects "abc", "3.5" and "-5"; int(raw) > 0 rejects "0"
    if raw.isdecimal() and int(raw) > 0:
        quantity = int(raw)
        break                      # success: leave the loop
    print(f"Invalid input '{raw}'. Enter a positive whole number.")
else:
    # runs only if the loop finished WITHOUT break (all attempts used)
    print("Too many invalid attempts. Order not created.")

if quantity is not None:
    print(f"Order accepted: {product} x {quantity} (quantity type: {type(quantity).__name__})") 

print("TASK 6: CATALOGUE SEARCH")

# Product catalogue from the RetailPulse orders
catalogue = ["AIRFRYER", "ELECTRIC KETTLE", "MICROWAVE OVEN", "REFRIGERATOR", "SMART TV", "LAPTOP"]

search = input("Enter product name to search: ").strip().upper()

for position, item in enumerate(catalogue, start=1):
    if item == search:
        print(f"{item} found at position {position}")
        break                      # found: stop searching
else:
    # runs only if the loop finished WITHOUT break (no match)
    print("Not stocked")
```
